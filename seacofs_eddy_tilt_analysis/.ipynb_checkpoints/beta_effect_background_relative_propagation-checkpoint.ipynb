# Beta effect after removing background EAC advection

This notebook asks whether EAC eddies have a polarity-dependent **residual drift** after subtracting the strong southwestward background current. It extends `beta_effect_vorticity_tilt_analysis.ipynb`.

The primary diagnostic is

$$\mathbf c_{res}=\mathbf c_{track}-\mathbf U_{background},$$

with the Southern Hemisphere beta-effect prediction

$$c_{res,N}>0\quad\mathrm{for\ AEs},\qquad c_{res,N}<0\quad\mathrm{for\ CEs}.$$

Four background estimates are compared: instantaneous surface annulus, instantaneous 0–500 m annulus, 26-year monthly surface climatology, and 26-year monthly 0–500 m climatology. The instantaneous annulus is primary; the others are sensitivity analyses. All velocities are true east/north.

Planetary-beta dominance retains the existing topographic PV term based on `w + f`. This deliberately allows AEs and CEs to experience different topographic-beta strengths at the same location. Consequently, dominance is evaluated separately for every eddy-day and polarity rather than using a common geographic mask.


```python
from pathlib import Path

import netCDF4 as nc
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf
from scipy.stats import spearmanr
from statsmodels.genmod.cov_struct import Exchangeable
from statsmodels.genmod.families import Gaussian

import seacofs_tilt_tools as tilt

pd.set_option("display.max_columns", 120)
plt.rcParams["figure.dpi"] = 120

MODEL_ROOT = Path("/srv/scratch/z3533156/26year_BRAN2020")
CACHE_ROOT = Path("/srv/scratch/z5297792/SEACOFS_26yr_eddy_dataset/background_flow_cache")
CLIM_CACHE = CACHE_ROOT / "monthly_surface_500m_climatology.npz"
EDDY_BACKGROUND_CACHE = CACHE_ROOT / "eddy_day_background_velocity.parquet"
DATE_ORIGIN = pd.Timestamp("1990-01-01")
DEPTH_MAX_M = 500.0
ANNULUS_INNER_RC = 1.5
ANNULUS_OUTER_RC = 3.0
MIN_ANNULUS_POINTS = 20
SMOOTH_DAYS = 5
N_BOOT = 2000
RANDOM_STATE = 42
SECONDS_PER_DAY = 86400.0
KM_TO_M = 1000.0

# Set False only when inspecting cached results without access to all model files.
BUILD_MISSING_CACHES = True
```

## 1. Load eddies and calculate polarity-specific PV dominance

`tilt.add_pv_gradient_terms` calculates

$$|\nabla q_{topo}|\propto |w+f|\,|\nabla h|/h^2,$$

so AE and CE dominance masks can differ. `topo_plan_ratio < 0` means the planetary PV-gradient magnitude exceeds the topographic magnitude. Threshold sensitivity is assessed later rather than replacing this physical definition.


```python
paths = tilt.Paths()
grid = tilt.load_grid(paths.grid, paths.z_r)
df_eddies, df_tilt = tilt.load_tilt_tables(paths)

required = {"Eddy", "Day", "Cyc", "xc", "yc", "ic", "jc",
            "w", "Rc", "TiltDis", "TiltDir"}
missing = sorted(required.difference(df_eddies.columns))
if missing:
    raise KeyError(f"Missing required columns: {missing}")
if df_eddies.duplicated(["Eddy", "Day"]).any():
    raise ValueError("Expected one row per Eddy-Day.")

df = tilt.add_pv_gradient_terms(df_eddies, grid)
df = df[df.Cyc.isin(["AE", "CE"])].sort_values(["Eddy", "Day"]).copy()
df["Date"] = DATE_ORIGIN + pd.to_timedelta(df["Day"].astype(int), unit="D")
df["month"] = df["Date"].dt.month
df["planetary_dominant"] = df["topo_plan_ratio"] < 0
df["planetary_to_topographic"] = np.exp(-df["topo_plan_ratio"])
df["age_days"] = df["Day"] - df.groupby("Eddy")["Day"].transform("min")

sign_check = df.groupby("Cyc").agg(
    rows=("Eddy", "size"), eddies=("Eddy", "nunique"),
    median_w=("w", "median"),
    planetary_dominant_fraction=("planetary_dominant", "mean"),
)
display(sign_check)
assert sign_check.loc["AE", "median_w"] > 0, "Expected positive AE relative vorticity."
assert sign_check.loc["CE", "median_w"] < 0, "Expected negative CE relative vorticity."
```

## 2. True east/north track velocity

DOPPIO `xc`, `yc` are kilometres along the model grid. With ROMS `angle` measured counter-clockwise from true east to the grid xi-axis, the inverse rotation is

$$c_E=c_x\cos\alpha-c_y\sin\alpha,\qquad c_N=c_x\sin\alpha+c_y\cos\alpha.$$

The diagnostic below checks that this agrees with longitude/latitude changes when those columns are available.


```python
def smooth_and_rate(part, column, smooth_days=SMOOTH_DAYS):
    part = part.sort_values("Day")
    time = part["Day"].to_numpy(float)
    smooth = (part[column].astype(float).rolling(
        smooth_days, center=True, min_periods=3).median().to_numpy())
    rate = np.full(len(part), np.nan)
    valid = np.isfinite(time) & np.isfinite(smooth)
    if valid.sum() >= 3 and np.all(np.diff(time[valid]) > 0):
        rate[valid] = np.gradient(smooth[valid], time[valid])
        valid_indices = np.flatnonzero(valid)
        for pos in valid_indices:
            neighbour_gaps = []
            if pos > 0 and valid[pos - 1]:
                neighbour_gaps.append(time[pos] - time[pos - 1])
            if pos + 1 < len(part) and valid[pos + 1]:
                neighbour_gaps.append(time[pos + 1] - time[pos])
            if not neighbour_gaps or max(neighbour_gaps) > 2 * smooth_days:
                rate[pos] = np.nan
    return pd.DataFrame({f"{column}_smooth": smooth, f"d{column}_day": rate}, index=part.index)

for column in ("xc", "yc", "w", "TiltDis"):
    derived = pd.concat([smooth_and_rate(part, column)
                         for _, part in df.groupby("Eddy", sort=False)]).sort_index()
    df = df.join(derived)

alpha = float(grid.angle)
df["track_east_km_day"] = df["dxc_day"] * np.cos(alpha) - df["dyc_day"] * np.sin(alpha)
df["track_north_km_day"] = df["dxc_day"] * np.sin(alpha) + df["dyc_day"] * np.cos(alpha)
df["dw_dt"] = df["dw_day"] / SECONDS_PER_DAY

if {"lon", "lat"}.issubset(df.columns):
    for column in ("lon", "lat"):
        derived = pd.concat([smooth_and_rate(part, column)
                             for _, part in df.groupby("Eddy", sort=False)]).sort_index()
        df = df.join(derived)
    df["track_north_from_lat_km_day"] = df["dlat_day"] * 111.195
    df["track_east_from_lon_km_day"] = (
        df["dlon_day"] * 111.195 * np.cos(np.deg2rad(df["lat_smooth"])))
    display(df[["track_east_km_day", "track_east_from_lon_km_day",
                "track_north_km_day", "track_north_from_lat_km_day"]].corr())
else:
    print("No lon/lat columns: using the ROMS-angle rotation of xc/yc.")
```

## 3. Background-flow estimators and cache construction

The instantaneous estimate is the robust median true-east/true-north flow in an annulus $1.5R_c<r<3R_c$. It captures the current state while excluding the fitted core. The monthly climatology averages all model days for each calendar month before sampling the closest grid point. The 0–500 m mean is a steering-depth sensitivity.

This section reads the 26-year NetCDF archive and can be expensive on its first run. Results are cached; later executions load the caches. Delete a cache deliberately if its configuration changes.


```python
def clean(values):
    values = np.asarray(values, float)
    return np.where(np.abs(values) > 1e30, np.nan, values)

depth_mask_500 = np.isfinite(grid.z_r) & (np.abs(grid.z_r) <= DEPTH_MAX_M)

def velocity_fields(dataset, t, include_depth_mean=True):
    # u_eastward/v_northward are already geographic; do not rotate them.
    u_surface = clean(dataset["u_eastward"][t, -1].T)
    v_surface = clean(dataset["v_northward"][t, -1].T)
    if not include_depth_mean:
        return u_surface, v_surface, None, None
    u3d = np.flip(clean(dataset["u_eastward"][t].T), axis=2)
    v3d = np.flip(clean(dataset["v_northward"][t].T), axis=2)
    with np.errstate(invalid="ignore"):
        u_500 = np.nanmean(np.where(depth_mask_500, u3d, np.nan), axis=2)
        v_500 = np.nanmean(np.where(depth_mask_500, v3d, np.nan), axis=2)
    return u_surface, v_surface, u_500, v_500

def model_file_for_day(day):
    fnumber = 1461 + ((int(day) - 1462) // 30) * 30
    return MODEL_ROOT / f"outer_avg_{fnumber:05d}.nc"

def build_monthly_climatology(output_path=CLIM_CACHE):
    files = sorted(MODEL_ROOT.glob("outer_avg_*.nc"))
    if not files:
        raise FileNotFoundError(f"No outer_avg files in {MODEL_ROOT}")
    shape = grid.X_grid.shape
    names = ("u_surface", "v_surface", "u_500", "v_500")
    sums = {name: np.zeros((12, *shape), dtype=np.float64) for name in names}
    counts = {name: np.zeros((12, *shape), dtype=np.int32) for name in names}
    for file_index, path in enumerate(files, 1):
        print(f"Climatology {file_index}/{len(files)}: {path.name}", end="\r")
        with nc.Dataset(path) as dataset:
            days = np.rint(clean(dataset["ocean_time"][:]) / SECONDS_PER_DAY).astype(int)
            for t, day in enumerate(days):
                month = (DATE_ORIGIN + pd.Timedelta(days=int(day))).month - 1
                fields = velocity_fields(dataset, t, include_depth_mean=True)
                for name, field in zip(names, fields):
                    valid = np.isfinite(field)
                    sums[name][month][valid] += field[valid]
                    counts[name][month][valid] += 1
    climatology = {name: np.divide(sums[name], counts[name],
                                  out=np.full_like(sums[name], np.nan),
                                  where=counts[name] > 0) for name in names}
    output_path.parent.mkdir(parents=True, exist_ok=True)
    np.savez_compressed(output_path, **climatology)
    print(f"\nSaved {output_path}")
    return climatology

def load_or_build_climatology():
    if CLIM_CACHE.exists():
        with np.load(CLIM_CACHE) as saved:
            return {name: saved[name] for name in saved.files}
    if not BUILD_MISSING_CACHES:
        raise FileNotFoundError(CLIM_CACHE)
    return build_monthly_climatology()
```

```python
def annulus_median(field, xc, yc, rc):
    if not np.isfinite(rc) or rc <= 0:
        return np.nan
    distance = np.hypot(grid.X_grid - xc, grid.Y_grid - yc)
    use = ((distance >= ANNULUS_INNER_RC * rc) &
           (distance <= ANNULUS_OUTER_RC * rc) & np.isfinite(field))
    if use.sum() < MIN_ANNULUS_POINTS:
        return np.nan
    values = field[use]
    low, high = np.nanpercentile(values, [10, 90])
    trimmed = values[(values >= low) & (values <= high)]
    return float(np.nanmedian(trimmed))

def build_eddy_background_table(eddy_table, climatology, output_path=EDDY_BACKGROUND_CACHE):
    rows = []
    day_groups = {int(day): part for day, part in eddy_table.groupby("Day", sort=False)}
    file_groups = {}
    for day in day_groups:
        file_groups.setdefault(model_file_for_day(day), []).append(day)
    for file_index, (path, days) in enumerate(sorted(file_groups.items()), 1):
        if not path.exists():
            print(f"Missing {path}")
            continue
        print(f"Eddy backgrounds {file_index}/{len(file_groups)}: {path.name}", end="\r")
        with nc.Dataset(path) as dataset:
            available = np.rint(clean(dataset["ocean_time"][:]) / SECONDS_PER_DAY).astype(int)
            lookup = {day: t for t, day in enumerate(available)}
            for day in days:
                t = lookup.get(day)
                if t is None:
                    continue
                u_s, v_s, u_500, v_500 = velocity_fields(dataset, t, include_depth_mean=True)
                month = (DATE_ORIGIN + pd.Timedelta(days=day)).month - 1
                for row in day_groups[day].itertuples(index=False):
                    ic, jc = int(row.ic), int(row.jc)
                    rows.append({
                        "Eddy": row.Eddy, "Day": day,
                        "ann_surface_east_ms": annulus_median(u_s, row.xc, row.yc, row.Rc),
                        "ann_surface_north_ms": annulus_median(v_s, row.xc, row.yc, row.Rc),
                        "ann_500_east_ms": annulus_median(u_500, row.xc, row.yc, row.Rc),
                        "ann_500_north_ms": annulus_median(v_500, row.xc, row.yc, row.Rc),
                        "clim_surface_east_ms": climatology["u_surface"][month, ic, jc],
                        "clim_surface_north_ms": climatology["v_surface"][month, ic, jc],
                        "clim_500_east_ms": climatology["u_500"][month, ic, jc],
                        "clim_500_north_ms": climatology["v_500"][month, ic, jc],
                    })
    result = pd.DataFrame(rows)
    output_path.parent.mkdir(parents=True, exist_ok=True)
    result.to_parquet(output_path, index=False)
    print(f"\nSaved {output_path}")
    return result

climatology = load_or_build_climatology()
if EDDY_BACKGROUND_CACHE.exists():
    background = pd.read_parquet(EDDY_BACKGROUND_CACHE)
elif BUILD_MISSING_CACHES:
    background = build_eddy_background_table(df, climatology)
else:
    raise FileNotFoundError(EDDY_BACKGROUND_CACHE)

if background.duplicated(["Eddy", "Day"]).any():
    raise ValueError("Background cache contains duplicate Eddy-Day rows.")
df = df.merge(background, on=["Eddy", "Day"], how="left", validate="one_to_one")
display(df[[column for column in df if column.endswith("_ms")]].describe().T)
```

## 4. Residual propagation under each steering estimate

Track velocity is converted from km day$^{-1}$ to m s$^{-1}$ before subtraction. Positive north is geographic north. `aligned_residual_north` is positive when the residual is equatorward for AEs or poleward for CEs.


```python
METHODS = {
    "ann_surface": ("ann_surface_east_ms", "ann_surface_north_ms"),
    "ann_500": ("ann_500_east_ms", "ann_500_north_ms"),
    "clim_surface": ("clim_surface_east_ms", "clim_surface_north_ms"),
    "clim_500": ("clim_500_east_ms", "clim_500_north_ms"),
}
df["track_east_ms"] = df["track_east_km_day"] * KM_TO_M / SECONDS_PER_DAY
df["track_north_ms"] = df["track_north_km_day"] * KM_TO_M / SECONDS_PER_DAY
df["expected_sign"] = df.Cyc.map({"AE": 1.0, "CE": -1.0}).astype(float)

for method, (east_background, north_background) in METHODS.items():
    df[f"{method}_residual_east_ms"] = df["track_east_ms"] - df[east_background]
    df[f"{method}_residual_north_ms"] = df["track_north_ms"] - df[north_background]
    df[f"{method}_aligned_residual_north_ms"] = (
        df["expected_sign"] * df[f"{method}_residual_north_ms"])

coverage = []
for method in METHODS:
    for cyc in ("AE", "CE"):
        part = df[df.Cyc == cyc]
        coverage.append({"method": method, "Cyc": cyc,
                         "rows": part[f"{method}_residual_north_ms"].notna().sum(),
                         "eddies": part.loc[part[f"{method}_residual_north_ms"].notna(), "Eddy"].nunique()})
display(pd.DataFrame(coverage).set_index(["method", "Cyc"]))
```

## 5. Primary planetary-beta-dominant pathway test

The primary population is `topo_plan_ratio < 0`, retaining the polarity-specific `w + f` calculation. Each eddy contributes one median residual velocity. Confidence intervals resample whole eddies. Raw true-north track velocity is included to show what background subtraction changes.


```python
def bootstrap_median(values, n_boot=N_BOOT, seed=RANDOM_STATE):
    values = np.asarray(values, float)
    values = values[np.isfinite(values)]
    rng = np.random.default_rng(seed)
    boot = np.array([np.nanmedian(rng.choice(values, len(values), replace=True))
                     for _ in range(n_boot)])
    return len(values), np.nanmedian(values), np.nanpercentile(boot, 2.5), np.nanpercentile(boot, 97.5)

primary = df[df.planetary_dominant].copy()
test_columns = {"raw_track": "track_north_ms",
                **{method: f"{method}_residual_north_ms" for method in METHODS}}
pathway_rows = []
eddy_pathways = {}
for method, column in test_columns.items():
    summary = primary.groupby(["Cyc", "Eddy"], as_index=False)[column].median().dropna()
    eddy_pathways[method] = summary
    for offset, cyc in enumerate(("AE", "CE")):
        part = summary[summary.Cyc == cyc].copy()
        n, estimate, low, high = bootstrap_median(part[column], seed=RANDOM_STATE + offset)
        expected = part[column] * (1 if cyc == "AE" else -1)
        pathway_rows.append({"method": method, "Cyc": cyc, "eddies": n,
                             "median_north_ms": estimate, "CI_low": low, "CI_high": high,
                             "expected_direction_fraction": np.mean(expected > 0)})
pathway_results = pd.DataFrame(pathway_rows)
display(pathway_results.set_index(["method", "Cyc"]).round(4))

fig, axes = plt.subplots(2, 2, figsize=(11, 8), sharex=True, sharey=True)
for ax, method in zip(axes.ravel(), METHODS):
    column = f"{method}_residual_north_ms"
    for cyc, color in (("AE", "darkred"), ("CE", "navy")):
        values = eddy_pathways[method].loc[eddy_pathways[method].Cyc == cyc, column]
        ax.hist(values, bins=35, histtype="step", lw=1.8, color=color, label=cyc)
    ax.axvline(0, color="k", lw=1)
    ax.set(title=method, xlabel="Residual northward velocity (m s$^{-1}$)", ylabel="Eddies")
axes[0, 0].legend(frameon=False)
fig.suptitle("Planetary-beta-dominant eddies: residual meridional propagation")
fig.tight_layout()
```

## 6. Residual-vorticity budget

For each background estimate, fit $d\zeta/dt\sim-\beta c_{res,N}$ separately by polarity with eddy-clustered robust uncertainty. A positive slope near one is the simplest absolute-vorticity expectation, but exact closure is not required because stretching and topographic PV terms remain.


```python
budget_rows = []
budget_models = {}
for method in METHODS:
    residual = f"{method}_residual_north_ms"
    primary[f"{method}_beta_tendency"] = -primary["beta"] * primary[residual]
    for cyc in ("AE", "CE"):
        fit = primary.loc[primary.Cyc == cyc, ["Eddy", "dw_dt", f"{method}_beta_tendency"]].dropna().copy()
        predictor = f"{method}_beta_tendency"
        scale = fit[predictor].std()
        fit["dw_scaled"] = fit["dw_dt"] / scale
        fit["beta_scaled"] = fit[predictor] / scale
        result = smf.gee("dw_scaled ~ beta_scaled", groups="Eddy", data=fit,
                         cov_struct=Exchangeable(), family=Gaussian()).fit()
        budget_models[(method, cyc)] = result
        ci = result.conf_int().loc["beta_scaled"]
        budget_rows.append({"method": method, "Cyc": cyc,
                            "rows": len(fit), "eddies": fit.Eddy.nunique(),
                            "slope": result.params["beta_scaled"],
                            "CI_low": ci.iloc[0], "CI_high": ci.iloc[1],
                            "intercept_scaled": result.params["Intercept"],
                            "observed_SD_s-2": fit.dw_dt.std(),
                            "beta_term_SD_s-2": fit[predictor].std()})
budget_results = pd.DataFrame(budget_rows)
display(budget_results.set_index(["method", "Cyc"]).round(4))
```

## 7. Does residual drift align with meridional tilt?

Positive `aligned_tilt` means equatorward for AEs and poleward for CEs. The within-eddy coefficient tests whether daily anomalies in beta-expected residual propagation accompany a more beta-aligned tilt after adjusting for age and radius.


```python
theta = np.deg2rad(primary["TiltDir"].astype(float))
primary["tilt_north_km"] = primary["TiltDis"] * np.cos(theta)
primary["aligned_tilt_km"] = primary["expected_sign"] * primary["tilt_north_km"]
alignment_rows = []
alignment_models = {}
for method in METHODS:
    source = f"{method}_aligned_residual_north_ms"
    for cyc in ("AE", "CE"):
        part = primary[primary.Cyc == cyc].copy()
        part["residual_between"] = part.groupby("Eddy")[source].transform("mean")
        part["residual_within"] = part[source] - part["residual_between"]
        for column in ("residual_between", "residual_within", "age_days", "Rc"):
            sd = part[column].std()
            part[f"{column}_z"] = (part[column] - part[column].mean()) / (sd if np.isfinite(sd) and sd > 0 else 1)
        needed = ["Eddy", "aligned_tilt_km", "residual_between_z",
                  "residual_within_z", "age_days_z", "Rc_z"]
        fit = part[needed].dropna().copy()
        result = smf.gee(
            "aligned_tilt_km ~ residual_between_z + residual_within_z + age_days_z + Rc_z",
            groups="Eddy", data=fit, cov_struct=Exchangeable(), family=Gaussian()).fit()
        alignment_models[(method, cyc)] = result
        ci = result.conf_int()
        alignment_rows.append({"method": method, "Cyc": cyc,
                               "between_coefficient_km": result.params["residual_between_z"],
                               "between_CI_low": ci.loc["residual_between_z", 0],
                               "between_CI_high": ci.loc["residual_between_z", 1],
                               "within_coefficient_km": result.params["residual_within_z"],
                               "within_CI_low": ci.loc["residual_within_z", 0],
                               "within_CI_high": ci.loc["residual_within_z", 1],
                               "rows": len(fit), "eddies": fit.Eddy.nunique()})
alignment_results = pd.DataFrame(alignment_rows)
display(alignment_results.set_index(["method", "Cyc"]).round(3))
```

## 8. Dominance and annulus sensitivity

The primary threshold is $|\nabla q_{planetary}|>|\nabla q_{topographic}|$. Stronger thresholds test whether the result sharpens as planetary beta becomes increasingly dominant. The existing `w + f` definition is retained throughout. Annulus-width sensitivity requires rebuilding only the eddy-day background cache with new constants.


```python
# ratio = planetary magnitude / topographic magnitude
DOMINANCE_THRESHOLDS = (0.5, 1.0, 2.0, 4.0)
threshold_rows = []
for ratio in DOMINANCE_THRESHOLDS:
    subset = df[df.planetary_to_topographic > ratio]
    for method in METHODS:
        column = f"{method}_residual_north_ms"
        eddies = subset.groupby(["Cyc", "Eddy"], as_index=False)[column].median().dropna()
        for cyc in ("AE", "CE"):
            values = eddies.loc[eddies.Cyc == cyc, column]
            n, estimate, low, high = bootstrap_median(values, seed=RANDOM_STATE + int(10 * ratio) + (cyc == "CE"))
            expected = values * (1 if cyc == "AE" else -1)
            threshold_rows.append({"planetary/topographic_threshold": ratio,
                                   "method": method, "Cyc": cyc, "eddies": n,
                                   "median_north_ms": estimate, "CI_low": low, "CI_high": high,
                                   "expected_direction_fraction": np.mean(expected > 0)})
threshold_results = pd.DataFrame(threshold_rows)
display(threshold_results.set_index(["planetary/topographic_threshold", "method", "Cyc"]).round(4))

fig, axes = plt.subplots(1, 2, figsize=(11, 4), sharey=True)
for ax, cyc in zip(axes, ("AE", "CE")):
    for method in METHODS:
        part = threshold_results[(threshold_results.Cyc == cyc) & (threshold_results.method == method)]
        aligned = part.median_north_ms * (1 if cyc == "AE" else -1)
        ax.plot(part["planetary/topographic_threshold"], aligned, "o-", label=method)
    ax.axhline(0, color="k", lw=1)
    ax.set(title=cyc, xlabel="Minimum planetary/topographic PV-gradient ratio",
           ylabel="Polarity-aligned residual north velocity (m s$^{-1}$)")
axes[0].legend(frameon=False, fontsize=8)
fig.tight_layout()
```

## 9. Interpretation checklist

The background-relative beta-effect interpretation is strongest if:

- raw tracks remain predominantly southwestward but AE residual north velocity becomes positive and CE residual north velocity remains negative;
- the result agrees across instantaneous annulus, monthly climatology, surface, and 0–500 m steering estimates;
- the expected polarity separation strengthens as the planetary/topographic PV-gradient ratio increases;
- $d\zeta/dt$ has a positive association with $-\beta c_{res,N}$, ideally with comparable magnitude;
- within-eddy residual propagation anomalies align with the observed meridional tilt component; and
- conclusions are stable to annulus boundaries, smoothing window, derivative outliers, and minimum track duration.

Disagreement among background estimates is itself informative: it indicates that the EAC steering flow is not cleanly separable from the eddy or is strongly depth-dependent. Do not select only the background estimate that gives the expected sign. Report all four, with eddy counts and whole-eddy uncertainty.
