```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

DATA_PATH = '/srv/scratch/z5297792/tilt_observation_data/MNF_IN2021_V03_adcp_trawler/MNF_IN2021_V03_adcp_trawler.csv'
```

```python
df = pd.read_csv(DATA_PATH)
df
```

```python
plt.plot(df['Mean Lon'], df['Mean Lat'])
```

```python
df.groupby('Depth').size().plot(marker='.')
DEPTH = df.Depth.unique()[0]
plt.xlabel('Depth (m)')
plt.ylabel('Number of observations')
plt.show()
```

```python
# df_eddy = df[(df['Mean Lon']>=153.75)&(df['Mean Lon']<=156.25)&(df['Mean Lat']<=-27)&(df['Mean Lat']>=-28)].copy()
df['Start (UTC)'] = pd.to_datetime(df['Start (UTC)'], utc=True)
df_eddy = df[
    (df['Start (UTC)'] >= pd.Timestamp('2021-05-8T23:30:00Z')) &
    (df['Start (UTC)'] <= pd.Timestamp('2021-05-09T14:30:00Z'))
].copy()
plt.plot(df_eddy['Mean Lon'], df_eddy['Mean Lat'])
df_eddy
```

```python
d = df_eddy.loc[df_eddy.Depth==DEPTH].copy()
t = mdates.date2num(pd.to_datetime(d['Start (UTC)']))

fig, ax = plt.subplots(figsize=(10, 10))
q = ax.quiver(d['Mean Lon'], d['Mean Lat'], d['U (m/sec)'], d['V (m/sec)'],
              t, cmap='viridis', scale=5)

cbar = fig.colorbar(q, ax=ax)
cbar.set_label('Time (UTC)')
cbar.ax.yaxis.set_major_formatter(mdates.DateFormatter('%H:%M %d/%m', tz=mdates.UTC))
cbar.update_ticks()

ticks = np.linspace(t.min(), t.max(), 6)
cbar.set_ticks(ticks)
cbar.set_ticklabels([mdates.num2date(x).strftime('%H:%M %d/%m') for x in ticks])

ax.set(xlabel='Longitude', ylabel='Latitude', aspect='equal')
plt.show()
```
